```python
def get_ac_from_row(host):
    if str(host['ra_frb']) == 'nan':
        hra = host['frb_ra']
        hdec = host['frb_dec']
    elif str(host['frb_ra']) == 'nan':
        hra = host['ra_frb']
        hdec = host['dec_frb']
    try:
        ra = float(hra)
        dec = float(hdec)
    except ValueError:
        if ':' in hra or 'h' in hra:
            hh,mm,ss =hra.split(':')
            ra = 15 * float(int(hh) + (int(mm) / 60) + (float(ss) / 3600))
        if ':' in hdec or 'h' in hdec:
            dd,mm,ss = hdec.split(':')
            dec = float(int(dd) + (int(mm) / 60) + (float(ss) / 3600))

    return ac.SkyCoord(ra = ra * u.deg, dec = dec * u.deg)
```
