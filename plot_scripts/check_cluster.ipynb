```python
plt.figure()
for iicat,cat in enumerate(all_cats[0:4]):
    plotcat = cat
    try:
        cat = cat[cat['z'] < 0.25]
    plt.scatter(plotcat['ra'],plotcat['dec'],marker = 'x')
    plt.scatter(hg.ra.degree,hg.dec.degree,color = f'C{iicat}',label = f'{iicat}')
    ax = plt.gca()
    for row in plotcat:
        # Calculate angular diameter distance
        ang_diam_dist = Planck18.angular_diameter_distance(row['z']).to(u.Mpc).value

        # Convert r200 to angular radius (assuming r200 is in kpc or Mpc)
        # Adjust the unit conversion as needed for your r200 units
        angular_radius = np.rad2deg(row['r500'] / ang_diam_dist) * (1/0.7) * 2

        # Create circle (radius in arcsec, converted to degrees for plotting)
        circle = Circle((row['ra'], row['dec']), 
                       angular_radius,  # convert arcsec to degrees
                       fill=False, edgecolor=f'C{iicat}', linewidth=1)
        ax.add_patch(circle)
        ax.text(row['ra'], row['dec'], f'z={row["z"]:.3f}', 
                ha='left', va='center', fontsize=8)

    ax.set_xlabel('RA (degrees)')
    ax.set_ylabel('Dec (degrees)')
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    plt.title('Wen+Han catalog')
    plt.show()
```
