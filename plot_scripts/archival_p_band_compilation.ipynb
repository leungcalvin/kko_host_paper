```python
cc = read_csv('/home/calvin/vlbi-correlator-localization/rampardath_and_others.csv')
cc.set_index(keys = 'name',inplace = True)
good = np.isfinite(cc['ra_decimal'] + cc['dec_decimal']) * (cc['peak_flux_jy'] > 1)

def mollweide_scatter(ax,ra,dec,**scatter_kwargs):
    ra_radians = np.deg2rad((np.atleast_1d(ra) + 180) % 360 - 180)
    dec_radians = np.deg2rad(np.atleast_1d(dec))
    circc = ax.scatter(ra_radians,dec_radians,**scatter_kwargs)
    return circc
fig = plt.figure(figsize = (10,6))
num_good = np.sum(good)
ax = fig.add_subplot(111, projection='mollweide')
plt.title(f'Confirmed <400 MHz; >1 Jy VLBA Calibrators\n (total = {num_good})')
plt.xlabel('Right ascension [deg]')
plt.ylabel('Declination [deg]')
for ref,marker in zip(
    ['rampadarath2009','lenc2008','chuprikov1999','clark1975'],
    ['o','x','v','^']):
    ref_good = cc['reference'][good].values
    ra_good = cc['ra_decimal'][good].values
    dec_good = cc['dec_decimal'][good].values
    flux_good = cc['peak_flux_jy'][good].values
    this_ref = (ref_good == ref)
    circc = mollweide_scatter(ax,ra_good[this_ref],
                              dec_good[this_ref],
                              marker = marker,
                              c = flux_good[this_ref],
                              label = ref,cmap = 'Spectral_r',
                              vmin = 1, vmax = 9)
plt.colorbar(circc,label = 'Compact Flux [Jy]')
plt.grid()
plt.legend(ncol = 2,loc = 'lower center')
plt.show()
```
