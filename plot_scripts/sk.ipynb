```python
import h5py
hsc_sk = h5py.File('/arc/home/calvin/kko_host_paper/data_products/suppression_pk_terasawa24_z00.hdf5','r')
```

```python
#plt.style.use('/arc/home/calvin/kko_host_paper/plot_scripts/paper_sty.mplstyle')
#plt.style.use('dark_background')
plt.figure(figsize = (4.4,3))
plt.xlabel(r'k (h Mpc$^{-1}$)')
plt.ylabel(r'S(k)')
plt.xlim(1e-1,15)
plt.ylim(0.7,1)

for sim,label,color in zip(['astrid','illustrisTNG','SIMBA',],['ASTRID','IllustrisTNG','SIMBA'],['C0','C1','C2']):
    kpk_dmo = np.loadtxt(f'/arc/home/calvin/kko_host_paper/data_products/{sim}_DM_Pk_m_z=0.05.txt')
    kpk_hydro = np.loadtxt(f'/arc/home/calvin/kko_host_paper/data_products/{sim}_Pk_m_z=0.05.txt')
    assert (kpk_dmo[:,0] == kpk_hydro[:,0]).all()
    sk = kpk_hydro[:,1] / kpk_dmo[:,1]
    plt.semilogx(kpk_dmo[:,0], sk,label = label,linewidth = 1.5 + (sim=='astrid'),color = color)
    plt.legend(bbox_to_anchor = [0,0.22,0.3,0.4],loc = 'upper left')
    plt.savefig(f'/arc/home/calvin/kko_host_paper/figures/sk_{sim}.pdf',dpi=100, bbox_inches='tight')

# ADD HSC Y3
#plt.semilogx(hsc_sk['k'][:], 0.5 * (hsc_sk['68lowerbounds'][:] + hsc_sk['68upperbounds'][:]),color = 'C5')
plt.fill_between(hsc_sk['k'][:],y1 = hsc_sk['68lowerbounds'][:],y2 = hsc_sk['68upperbounds'][:],alpha = 0.5,label = 'HSC Y3' ,color = 'C7')

# ADD DES Y3
#plt.semilogx(data['kk'],data['des_mean_sk'],color = 'C3')
plt.fill_between(data['kk'],data['des_lower_68_sk'],data['des_upper_68_sk'],alpha = 0.5,label = 'DES Y3',color = 'C3')
plt.legend(bbox_to_anchor = [0,0.22,0.3,0.4],loc = 'upper left')
plt.savefig(f'/arc/home/calvin/kko_host_paper/figures/sk_wl.pdf',dpi=100, bbox_inches='tight')

kpk_dmo = np.loadtxt(f'/arc/home/calvin/kko_host_paper/data_products/astrid_DM_Pk_m_z=0.05.txt')
kpk_hydro = np.loadtxt(f'/arc/home/calvin/kko_host_paper/data_products/astrid_Pk_m_z=0.05.txt')
sk = kpk_hydro[:,1] / kpk_dmo[:,1]

plt.fill_between(x = kpk_dmo[:,0], 
             y1 = sk,
             y2 = 1, 
             hatch='xx', 
             facecolor='C0',edgecolor = 'red', label = '~20 FRBs',alpha = 0.8)
plt.legend(bbox_to_anchor = [0,0.22,0.3,0.4],loc = 'upper left')
plt.savefig(f'/arc/home/calvin/kko_host_paper/figures/sk_frb.pdf',dpi=100, bbox_inches='tight')




```
